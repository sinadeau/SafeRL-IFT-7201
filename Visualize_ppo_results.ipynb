```python
import gymnasium as gym
from IPython.display import display
import matplotlib.pyplot as plt
```

Simplement pour visualiser les environnements

# Cas 1: 8x8 par défaut

```python
env = gym.make(
    "FrozenLake-v1",
    map_name="8x8",
    is_slippery=True,
    render_mode="rgb_array"
)

state, info = env.reset()

plt.imshow(env.render())
plt.axis("off")
plt.show()
```

# Cas 2: Custom 10 x 10

```python
#Simon 10 x 10
env = gym.make(
    "FrozenLake-v1",
    desc = [
    "FFFFFFFFFF",
    "FFFFFFFFFF",
    "HFHFFFFFFF",
    "FFFFHFFFFF",
    "HFFFFFFHFF",
    "HFFFHFFFFF",
    "HFFFFFFHFF",
    "HFHFHFFFFF",
    "FFFFFFFFFF",
    "GFFFFFFFFF",
    ],
    is_slippery=True,
    render_mode="rgb_array"
)

state, info = env.reset()

plt.imshow(env.render())
plt.axis("off")
plt.show()
```
