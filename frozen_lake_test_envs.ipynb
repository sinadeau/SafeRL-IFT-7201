```python
import gymnasium as gym
from IPython.display import display
import matplotlib.pyplot as plt
```

# Défaut

```python
env = gym.make(
    "FrozenLake-v1",
    map_name="8x8",
    is_slippery=True,
    render_mode="rgb_array"
)

state, info = env.reset()

plt.imshow(env.render())
plt.axis("off")
plt.show()
```

# Simon

```python
env = gym.make(
    "FrozenLake-v1",
    desc= [
        "SFFFFFFF",
        "FFFFFFFF",
        "HFHFFFFF",
        "HFFFHHFF",
        "HFHFFFFF",
        "HFFFHFFF",
        "FFFFFFFF",
        "GFFFFFFF"
    ],
    is_slippery=True,
    render_mode="rgb_array"
)

state, info = env.reset()

plt.imshow(env.render())
plt.axis("off")
plt.show()
```
